```python
import georinex as gr
import numpy as np
import pandas as pd

# Membaca file RINEX BAKO DOY 305
obs = gr.load(rinex_file)

# Mengambil seluruh timestamp observasi
timestamps = pd.to_datetime(obs.time.values)

# Menghitung interval antar-epoch
intervals = timestamps.to_series().diff().dt.total_seconds().dropna()

print("=" * 60)
print("VERIFIKASI EPOCH AKTUAL BAKO DOY 305")
print("=" * 60)

print("Epoch pertama :", timestamps[0])
print("Epoch terakhir:", timestamps[-1])
print("Jumlah epoch  :", len(timestamps))

print("\nINTERVAL PENGAMATAN")
print("Median interval:", intervals.median(), "detik")
print("Interval minimum:", intervals.min(), "detik")
print("Interval maksimum:", intervals.max(), "detik")

print("\nINFORMASI DIMENSI DATA")
print(obs.sizes)
```

```python
import hashlib

def calculate_sha256(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            sha256.update(chunk)

    return sha256.hexdigest()

checksum = calculate_sha256(extracted_file)

print("CHECKSUM SHA-256 BAKO DOY 305")
print("=" * 60)
print("File   :", extracted_file.name)
print("SHA-256:", checksum)
```

```python
import unlzw3

# Lokasi arsip asli
compressed_file = DATA_DIR / "bako3050.03o.Z"

# Lokasi file hasil ekstraksi sementara
comparison_file = DATA_DIR / "bako3050_pembanding.03o"

# Membaca dan mengekstrak arsip .Z
with open(compressed_file, "rb") as f:
    compressed_data = f.read()

extracted_data = unlzw3.unlzw(compressed_data)

# Menyimpan hasil ekstraksi ke file pembanding
with open(comparison_file, "wb") as f:
    f.write(extracted_data)

print("EKSTRAKSI ARSIP SELESAI")
print("=" * 60)
print("Arsip asli       :", compressed_file.name)
print("File pembanding  :", comparison_file.name)
print("Ukuran hasil     :", comparison_file.stat().st_size, "bytes")
print("File berhasil dibuat:", comparison_file.exists())
```

```python
# Menghitung checksum file hasil ekstraksi ulang
checksum_pembanding = calculate_sha256(comparison_file)

# Membandingkan dengan checksum file awal
print("PERBANDINGAN CHECKSUM BAKO DOY 305")
print("=" * 70)

print("File awal:")
print(calculate_sha256(extracted_file))

print("\nFile hasil ekstraksi ulang:")
print(checksum_pembanding)

print("\nHASIL PERBANDINGAN:")
if calculate_sha256(extracted_file) == checksum_pembanding:
    print("IDENTIK — kedua file memiliki isi yang sama.")
else:
    print("BERBEDA — kedua file memiliki isi yang berbeda.")
```
