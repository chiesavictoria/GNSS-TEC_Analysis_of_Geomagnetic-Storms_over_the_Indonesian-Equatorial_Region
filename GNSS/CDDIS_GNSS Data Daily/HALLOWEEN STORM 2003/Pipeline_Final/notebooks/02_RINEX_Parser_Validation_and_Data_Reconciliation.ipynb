# PIPELINE 02 — RINEX Parser Validation and Data Reconciliation

## 1. Overview

Pipeline 02 merupakan tahap validasi dan rekonsiliasi hasil parsing data GNSS RINEX yang telah dikembangkan pada Pipeline 01.

Pipeline ini bertujuan untuk mengevaluasi kesesuaian hasil parser manual terhadap struktur dan isi file RINEX mentah. Validasi dilakukan melalui pemeriksaan struktur epoch, identitas satelit, pemetaan observasi, serta perbandingan dengan hasil pembacaan independen menggunakan GeoRinex.

Dataset utama yang digunakan pada tahap awal adalah data observasi GNSS stasiun BAKO tanggal 27 Oktober 2003 (DOY300), dalam format RINEX 2.10.

## 2. Objectives

Tujuan Pipeline 02 adalah:

1. Memeriksa integritas file input dan hasil parsing dari Pipeline 01.
2. Memvalidasi kesesuaian struktur epoch antara file RINEX mentah dan hasil parser.
3. Memvalidasi identitas serta jumlah satelit pada setiap epoch.
4. Memeriksa ketepatan pemetaan nilai observasi berdasarkan satelit dan tipe observasi.
5. Memvalidasi penanganan missing values, Loss of Lock Indicator (LLI), dan Signal Strength Indicator (SSI).
6. Membandingkan hasil parser manual dengan hasil pembacaan GeoRinex.
7. Mengidentifikasi dan mendokumentasikan ketidaksesuaian yang ditemukan.
8. Menetapkan status validasi parser berdasarkan kriteria yang telah ditentukan.

## 3. Scope and Limitations

### 3.1 Scope

Pengerjaan Pipeline 02 mencakup:

- Validasi input dan output Pipeline 01.
- Pemeriksaan ulang struktur RINEX mentah.
- Rekonsiliasi data pada tingkat epoch.
- Rekonsiliasi identitas satelit.
- Rekonsiliasi nilai observasi.
- Perbandingan hasil parser manual dengan GeoRinex.
- Identifikasi dan klasifikasi discrepancy.
- Penyusunan laporan validasi.

### 3.2 Limitations

Batasan pengerjaan ditetapkan sebagai berikut:

- Dataset utama yang divalidasi adalah BAKO DOY300.
- Format data adalah RINEX 2.10 GPS Observation.
- Interval observasi nominal adalah 30 detik.
- Tipe observasi yang diperiksa adalah L1, L2, C1, P2, P1, dan D1.
- Parser manual yang digunakan berasal dari Pipeline 01.
- GeoRinex digunakan sebagai pembanding independen, bukan sebagai acuan kebenaran absolut.
- Basis waktu GPS dipertahankan selama proses validasi. Konversi ke UTC harus dilakukan secara eksplisit apabila diperlukan.
- Missing values tetap dipertahankan sebagai nilai kosong atau NaN.
- Data mentah tidak dimodifikasi.
- Ketidaksesuaian tidak dikoreksi secara otomatis tanpa pemeriksaan dan dokumentasi.
- Kesimpulan validasi dibatasi pada dataset dan aspek yang benar-benar telah diuji.

### 3.3 Excluded Activities

Aktivitas berikut tidak termasuk dalam ruang lingkup Pipeline 02:

- Perhitungan STEC dan VTEC.
- Koreksi cycle slip.
- Deteksi cycle slip secara menyeluruh.
- Segmentasi satellite arc.
- Koreksi satellite bias dan receiver bias.
- Analisis respons TEC terhadap indeks geomagnetik.
- Analisis komparatif badai geomagnetik.

Aktivitas tersebut akan dikerjakan pada pipeline berikutnya sesuai dengan tahapan penelitian.

````python
# ============================================================
# 02.03.02 - RAW RINEX HEADER METADATA EXTRACTION
# ============================================================

header_metadata = {}

for line in raw_header_lines:
    label = line[60:80].strip() if len(line) >= 60 else ""

    if label == "RINEX VERSION / TYPE":
        header_metadata["RINEX_Version"] = line[0:9].strip()
        header_metadata["File_Type"] = line[20:40].strip()
        header_metadata["Satellite_System"] = line[40:60].strip()

    elif label == "MARKER NAME":
        header_metadata["Marker_Name"] = line[0:60].strip()

    elif label == "MARKER NUMBER":
        header_metadata["Marker_Number"] = line[0:20].strip()

    elif label == "REC # / TYPE / VERS":
        header_metadata["Receiver_Number"] = line[0:20].strip()
        header_metadata["Receiver_Type"] = line[20:40].strip()
        header_metadata["Receiver_Version"] = line[40:60].strip()

    elif label == "ANT # / TYPE":
        header_metadata["Antenna_Number"] = line[0:20].strip()
        header_metadata["Antenna_Type"] = line[20:40].strip()

    elif label == "APPROX POSITION XYZ":
        header_metadata["Approx_Position_XYZ"] = line[0:60].strip()

    elif label == "ANTENNA: DELTA H/E/N":
        header_metadata["Antenna_Delta_HEN"] = line[0:60].strip()

    elif label == "# / TYPES OF OBSERV":
        header_metadata["Observation_Types_Count"] = int(line[0:6])
        header_metadata["Observation_Types"] = [
            line[i:i+6].strip()
            for i in range(6, 60, 6)
            if line[i:i+6].strip()
        ]

    elif label == "TIME OF FIRST OBS":
        header_metadata["First_Observation_Time"] = line[0:43].strip()
        header_metadata["Time_System"] = line[48:51].strip()

# Tabel metadata
df_raw_header_metadata = pd.DataFrame(
    [
        {"Metadata": key, "Value": value}
        for key, value in header_metadata.items()
    ]
)

display(df_raw_header_metadata)

# Validasi jumlah tipe observasi
declared_count = header_metadata["Observation_Types_Count"]
extracted_count = len(header_metadata["Observation_Types"])

print("\nDeclared observation types :", declared_count)
print("Extracted observation types:", extracted_count)
print("Observation type count matches:", declared_count == extracted_count)

# Simpan hasil
raw_reference_dir = OUTPUT_DIR / "raw_reference"
raw_reference_dir.mkdir(parents=True, exist_ok=True)

metadata_output = (
    raw_reference_dir / "raw_header_metadata_doy300.csv"
)

df_raw_header_metadata.to_csv(metadata_output, index=False)

print("\nOutput saved:")
print(metadata_output)
````

````python
# ============================================================
# 02.03.03 - RAW RINEX EPOCH RECORD IDENTIFICATION
# ============================================================

import re

raw_epoch_records = []

for line_index in range(HEADER_END_INDEX + 1, len(raw_lines)):
    line = raw_lines[line_index]

    if not line.startswith(" "):
        continue

    if len(line) < 32:
        continue

    try:
        year = int(line[1:3])
        month = int(line[4:6])
        day = int(line[7:9])
        hour = int(line[10:12])
        minute = int(line[13:15])
        second = float(line[15:26])
        epoch_flag = int(line[28:29])
        declared_satellites = int(line[29:32])

    except ValueError:
        continue

    if not (
        1 <= month <= 12
        and 1 <= day <= 31
        and 0 <= hour <= 23
        and 0 <= minute <= 59
        and 0 <= second < 61
    ):
        continue

    year = 2000 + year if year < 80 else 1900 + year

    timestamp = (
        pd.Timestamp(
            year=year,
            month=month,
            day=day,
            hour=hour,
            minute=minute
        )
        + pd.Timedelta(seconds=second)
    )

    raw_epoch_records.append({
        "Raw_Line_Index": line_index,
        "Timestamp_GPS": timestamp,
        "Epoch_Flag": epoch_flag,
        "Declared_Satellites": declared_satellites,
        "Raw_Epoch_Line": line.rstrip("\r\n")
    })

df_raw_epochs = pd.DataFrame(raw_epoch_records)

print("=" * 70)
print("RAW RINEX EPOCH RECORD IDENTIFICATION")
print("=" * 70)

print("Total epoch records       :", len(df_raw_epochs))
print("Unique timestamps         :", df_raw_epochs["Timestamp_GPS"].nunique())
print("Epoch flag distribution:")
print(df_raw_epochs["Epoch_Flag"].value_counts().sort_index().to_string())

print("\nDeclared satellites distribution:")
print(
    df_raw_epochs["Declared_Satellites"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nFirst 5 epoch records:")
display(df_raw_epochs.head())

print("\nLast 5 epoch records:")
display(df_raw_epochs.tail())
````

````python
# ============================================================
# 02.03.04 - SATELLITE LIST STRUCTURE INSPECTION
# ============================================================

satellite_list_results = []

for _, epoch in df_raw_epochs.iterrows():

    line_index = int(epoch["Raw_Line_Index"])
    declared_count = int(epoch["Declared_Satellites"])

    # Baris pertama daftar satelit
    first_line = raw_lines[line_index].rstrip("\r\n")

    satellite_text = first_line[32:68]

    # Ekstraksi ID satelit dari kolom 33-68
    satellites = [
        satellite_text[i:i+3].strip().replace(" ", "0")
        for i in range(0, len(satellite_text), 3)
        if satellite_text[i:i+3].strip()
    ]

    # Periksa kemungkinan baris lanjutan
    continuation_lines = 0
    next_index = line_index + 1

    while len(satellites) < declared_count and next_index < len(raw_lines):

        continuation_line = raw_lines[next_index].rstrip("\r\n")

        if continuation_line.startswith(" " * 32):
            continuation_text = continuation_line[32:68]

            additional_satellites = [
                continuation_text[i:i+3].strip().replace(" ", "0")
                for i in range(0, len(continuation_text), 3)
                if continuation_text[i:i+3].strip()
            ]

            satellites.extend(additional_satellites)
            continuation_lines += 1
            next_index += 1

        else:
            break

    satellite_list_results.append({
        "Timestamp_GPS": epoch["Timestamp_GPS"],
        "Declared_Satellites": declared_count,
        "Extracted_Satellites": len(satellites),
        "Satellite_IDs": satellites,
        "Continuation_Lines": continuation_lines,
        "Count_Matches": len(satellites) == declared_count
    })

df_satellite_structure = pd.DataFrame(satellite_list_results)

print("=" * 70)
print("SATELLITE LIST STRUCTURE INSPECTION")
print("=" * 70)

print("Total epochs:", len(df_satellite_structure))
print(
    "Epochs with matching satellite count:",
    df_satellite_structure["Count_Matches"].sum()
)
print(
    "Epochs with mismatched satellite count:",
    (~df_satellite_structure["Count_Matches"]).sum()
)

print("\nContinuation line distribution:")
print(
    df_satellite_structure["Continuation_Lines"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nSatellite count comparison:")
print(
    df_satellite_structure.groupby(
        ["Declared_Satellites", "Extracted_Satellites"]
    ).size().to_string()
)

print("\nFirst 5 records:")
display(df_satellite_structure.head())

print("\nMismatched records:")
display(
    df_satellite_structure[
        ~df_satellite_structure["Count_Matches"]
    ].head(10)
)
````

````python
# ============================================================
# 02.05.02 - RAW RINEX SATELLITE REFERENCE PREPARATION
# ============================================================

raw_satellite_records = []

for i, line in enumerate(raw_lines):

    # Lewati baris sebelum data epoch
    if len(line) < 32:
        continue

    # Identifikasi baris epoch berdasarkan format RINEX 2
    try:
        yy = int(line[1:3])
        month = int(line[4:6])
        day = int(line[7:9])
        hour = int(line[10:12])
        minute = int(line[13:15])
        second = float(line[15:26])

        if not (
            1 <= month <= 12
            and 1 <= day <= 31
            and 0 <= hour <= 23
            and 0 <= minute <= 59
            and 0 <= second < 61
        ):
            continue

        timestamp = (
            pd.Timestamp(
                year=2000 + yy,
                month=month,
                day=day,
                hour=hour,
                minute=minute
            )
            + pd.Timedelta(seconds=second)
        )

        declared_count = int(line[29:32])

    except (ValueError, TypeError):
        continue

    # Ambil daftar satelit pada baris epoch
    satellite_text = line[32:68]

    satellite_ids = [
        satellite_text[j:j+3].strip().replace(" ", "0")
        for j in range(0, len(satellite_text), 3)
        if satellite_text[j:j+3].strip()
    ]

    # Tangani continuation line jika daftar satelit melebihi baris pertama
    next_line = i + 1

    while len(satellite_ids) < declared_count:
        continuation = raw_lines[next_line].rstrip("\r\n")
        satellite_text = continuation[32:68]

        satellite_ids.extend([
            satellite_text[j:j+3].strip().replace(" ", "0")
            for j in range(0, len(satellite_text), 3)
            if satellite_text[j:j+3].strip()
        ])

        next_line += 1

    for satellite_id in satellite_ids:
        raw_satellite_records.append({
            "Timestamp_GPS": timestamp,
            "Satellite_ID": satellite_id
        })

df_raw_satellite_check = pd.DataFrame(raw_satellite_records)

df_raw_satellite_check = (
    df_raw_satellite_check
    .sort_values(["Timestamp_GPS", "Satellite_ID"])
    .reset_index(drop=True)
)

print("=" * 70)
print("RAW RINEX SATELLITE REFERENCE PREPARATION")
print("=" * 70)

print("Total epoch-satellite pairs:",
      len(df_raw_satellite_check))

print("Unique timestamps:",
      df_raw_satellite_check["Timestamp_GPS"].nunique())

print("Unique satellite IDs:",
      df_raw_satellite_check["Satellite_ID"].nunique())

print("Duplicate epoch-satellite pairs:",
      df_raw_satellite_check.duplicated(
          ["Timestamp_GPS", "Satellite_ID"]
      ).sum())

print("\nFirst five records:")
display(df_raw_satellite_check.head())
````

````python
# ============================================================
# 02.06.01 — RAW OBSERVATION RECORD EXTRACTION
# ============================================================

import math

# Observation types verified from the raw RINEX header
obs_types = ["L1", "L2", "C1", "P2", "P1", "D1"]

raw_observation_records = []

# Locate the end of the RINEX header
header_end_index = next(
    i for i, line in enumerate(raw_lines)
    if "END OF HEADER" in line
)

# RINEX 2 observation records: 5 fields per physical line
fields_per_line = 5
observation_lines_per_satellite = math.ceil(
    len(obs_types) / fields_per_line
)

i = header_end_index + 1

while i < len(raw_lines):

    epoch_line = raw_lines[i].rstrip("\r\n")

    # Read epoch header
    try:
        yy = int(epoch_line[1:3])
        month = int(epoch_line[4:6])
        day = int(epoch_line[7:9])
        hour = int(epoch_line[10:12])
        minute = int(epoch_line[13:15])
        second = float(epoch_line[15:26])
        epoch_flag = int(epoch_line[28])
        declared_count = int(epoch_line[29:32])

        if not (1 <= month <= 12 and 1 <= day <= 31):
            raise ValueError("Invalid epoch date")

    except (ValueError, IndexError):
        raise ValueError(
            f"Unable to parse epoch header at raw line {i + 1}: "
            f"{epoch_line!r}"
        )

    year = 1900 + yy if yy >= 80 else 2000 + yy

    timestamp = (
        pd.Timestamp(
            year=year,
            month=month,
            day=day,
            hour=hour,
            minute=minute
        )
        + pd.Timedelta(seconds=second)
    )

    # Read satellite IDs from epoch header
    satellite_ids = []

    satellite_text = epoch_line[32:68]

    satellite_ids.extend([
        satellite_text[j:j + 3].strip()
        for j in range(0, len(satellite_text), 3)
        if satellite_text[j:j + 3].strip()
    ])

    i += 1

    # Read satellite-list continuation lines if necessary
    while len(satellite_ids) < declared_count:

        if i >= len(raw_lines):
            raise ValueError(
                "Unexpected end of file while reading satellite list"
            )

        continuation = raw_lines[i].rstrip("\r\n")
        satellite_text = continuation[32:68]

        satellite_ids.extend([
            satellite_text[j:j + 3].strip()
            for j in range(0, len(satellite_text), 3)
            if satellite_text[j:j + 3].strip()
        ])

        i += 1

    if len(satellite_ids) != declared_count:
        raise ValueError(
            f"Satellite count mismatch at {timestamp}: "
            f"declared={declared_count}, extracted={len(satellite_ids)}"
        )

    # Read observation records for each satellite
    for satellite_id in satellite_ids:

        satellite_observation_lines = []

        for _ in range(observation_lines_per_satellite):

            if i >= len(raw_lines):
                raise ValueError(
                    "Unexpected end of file while reading observations"
                )

            satellite_observation_lines.append(
                raw_lines[i].rstrip("\r\n").ljust(80)
            )

            i += 1

        # Extract each observation field
        for obs_index, obs_type in enumerate(obs_types):

            line_offset = obs_index // fields_per_line
            field_index = obs_index % fields_per_line

            raw_field = satellite_observation_lines[
                line_offset
            ][field_index * 16:(field_index + 1) * 16]

            value_text = raw_field[0:14].strip()
            lli_text = raw_field[14:15].strip()
            ssi_text = raw_field[15:16].strip()

            value = float(value_text) if value_text else float("nan")
            lli = int(lli_text) if lli_text else float("nan")
            ssi = int(ssi_text) if ssi_text else float("nan")

            raw_observation_records.append({
                "Timestamp_GPS": timestamp,
                "Epoch_Flag": epoch_flag,
                "Satellite_ID": satellite_id,
                "Observation_Type": obs_type,
                "Value": value,
                "LLI": lli,
                "SSI": ssi,
                "Raw_Field": raw_field,
            })


# Create raw observation DataFrame
df_raw_observations = pd.DataFrame(raw_observation_records)

# Sort records for consistent comparison
df_raw_observations = (
    df_raw_observations
    .sort_values(
        ["Timestamp_GPS", "Satellite_ID", "Observation_Type"]
    )
    .reset_index(drop=True)
)

# ============================================================
# SUMMARY
# ============================================================

print("=" * 65)
print("RAW OBSERVATION RECORD EXTRACTION")
print("=" * 65)

print(f"Total observation records : {len(df_raw_observations)}")
print(
    "Unique timestamps         :",
    df_raw_observations["Timestamp_GPS"].nunique()
)
print(
    "Unique satellite IDs      :",
    df_raw_observations["Satellite_ID"].nunique()
)
print(
    "Observation types         :",
    df_raw_observations["Observation_Type"].nunique()
)
print(
    "Duplicate observation keys:",
    df_raw_observations.duplicated(
        ["Timestamp_GPS", "Satellite_ID", "Observation_Type"]
    ).sum()
)

print("\nMissing values:")
print(
    df_raw_observations[
        ["Value", "LLI", "SSI"]
    ].isna().sum()
)

print("\nRecords per observation type:")
print(
    df_raw_observations["Observation_Type"]
    .value_counts()
    .sort_index()
)

print("\nFirst five records:")
display(df_raw_observations.head())

# Save raw observation reference
raw_observation_output = (
    OUTPUT_DIR / "observation_reconciliation" /
    "raw_observation_reference_doy300.csv"
)

df_raw_observations.to_csv(
    raw_observation_output,
    index=False
)

print("\nOutput saved:")
print(raw_observation_output)
````

## 02.09 — Validation Assessment and Acceptance Criteria

### Objective

Evaluate the performance and reliability of the manually developed RINEX parser based on the reconciliation results obtained from the raw RINEX file and independent GeoRinEx processing.

### Validation Scope

The assessment is limited to:

- BAKO station, DOY300 (27 October 2003).
- RINEX version 2.10, GPS observation data.
- Six observation types: L1, L2, C1, P2, P1, and D1.
- Epoch-level, satellite-level, and observation-level consistency.
- Timestamp consistency and missing-value preservation.

### Acceptance Criteria

1. Epoch timestamps must match the original raw RINEX timestamps.
2. Epoch flags and declared satellite counts must be consistent.
3. Epoch-satellite pairs must match the raw RINEX records.
4. Observation keys must be complete and unique.
5. Available observation values must match the raw RINEX reference.
6. Missing-value patterns must be preserved.
7. Independent GeoRinEx comparison must show consistent observation keys and values.
8. All identified discrepancies must be documented and assessed.

### Assessment Categories

- **PASS:** The criterion is satisfied by the available validation evidence.
- **REVIEW:** The criterion requires further investigation or has a limitation that must be documented.
- **FAIL:** The criterion is not satisfied by the validation evidence.
